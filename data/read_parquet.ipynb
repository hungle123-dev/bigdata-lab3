# Task 2.2 Result Reader: Task_2-2.parquet
Notebook này được dùng để đọc, hiển thị và kiểm tra độ chính xác của file kết quả `Task_2-2.parquet` thu được từ Spark.

```python
import pandas as pd

# Đọc file Parquet cùng cấp thư mục
file_path = 'Task_2-2.parquet'
df = pd.read_parquet(file_path)

print(f"=== KÍCH THƯỚC DỮ LIỆU ===")
print(f"- Số nhóm (SKU, Month): {len(df):,}")
print(f"- Số cột dữ liệu: {len(df.columns)}")
print(f"\nDanh sách các cột trong file:\n{df.columns.tolist()}")
```

```python
# Hiển thị 20 mẫu kết quả đầu tiên
df.head(20)
```

```python
# Kiểm tra các nhóm có chỉ số StdDev > 0
non_zero_stddev = df[df['p90_stddev_exact'] > 0]
print(f"Số nhóm có StdDev P90 > 0: {len(non_zero_stddev):,} / {len(df):,}")
non_zero_stddev.head(10)
```

```python
# Kiểm tra tổng quan sai lệch giữa 2 phương pháp (Approximate vs Exact)
print("=== BÁO CÁO ĐỘ LỆCH (APPROXIMATE VS EXACT) ===")
p90_thresh_diff = (df['p90_thresh_approx'] - df['p90_thresh_exact']).abs().max()
p80_thresh_diff = (df['p80_thresh_approx'] - df['p80_thresh_exact']).abs().max()
p90_stddev_diff = (df['p90_stddev_approx'] - df['p90_stddev_exact']).abs().max()
p80_stddev_diff = (df['p80_stddev_approx'] - df['p80_stddev_exact']).abs().max()

print(f"- Sai lệch lớn nhất Ngưỡng P90 : {p90_thresh_diff:.6f}")
print(f"- Sai lệch lớn nhất Ngưỡng P80 : {p80_thresh_diff:.6f}")
print(f"- Sai lệch lớn nhất StdDev P90 : {p90_stddev_diff:.6f}")
print(f"- Sai lệch lớn nhất StdDev P80 : {p80_stddev_diff:.6f}")
```
